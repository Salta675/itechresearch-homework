# 2. Әртүрлі өндірушілердің өнімдерінің тағамдық құрамы қалай ерекшеленеді?

```python
nutrition = df.filter(items=[
    'name', 'mfr',
    'calories', 'protein', 'fat',
    'sodium', 'fiber', 'carbo',
    'sugars', 'potass', 'vitamins'
]).copy()
```

```python
nutrition.shape
```

```python
nutrition['mfr'] = nutrition['mfr'].map(manufacturers)
```

```python
nutrition.head()
```

```python
nutrition['sugar_level'] = 'Moderate'

nutrition.loc[nutrition['sugars'] <= 5, 'sugar_level'] = 'Low'
nutrition.loc[nutrition['sugars'] > 10, 'sugar_level'] = 'High'

nutrition['fiber_level'] = 'Moderate'

nutrition.loc[nutrition['fiber'] < 3, 'fiber_level'] = 'Low'
nutrition.loc[nutrition['fiber'] >= 5, 'fiber_level'] = 'High'

nutrition['protein_level'] = 'Moderate'

nutrition.loc[nutrition['protein'] < 3, 'protein_level'] = 'Low'
nutrition.loc[nutrition['protein'] >= 4, 'protein_level'] = 'High'
```

```python
nutrition.head()
```

```python
nutrition['sugar_level'].value_counts()
```

```python
nutrition.loc[
    nutrition['sugar_level'] == 'High',
    ['name', 'mfr', 'sugars', 'sugar_level']
].sort_values(by='sugars', ascending=False)
```

```python
nutrition[nutrition['sugar_level'] == 'High'].groupby('mfr')['sugar_level'].value_counts()
```

```python
nutrition[nutrition['sugar_level'] == 'Low'].groupby('mfr')['sugar_level'].value_counts()
```

# 3. Өнімнің тағамдық құрамы мен рейтингі арасында қандай байланыс бар?

```python
rating_df = df.filter(items=[
    'name', 'mfr', 'calories', 'sugars',
     'protein', 'fiber', 'fat', 'sodium', 'rating']
).copy()
```

```python
rating_df.head()
```

```python
rating_df['mfr'] = rating_df['mfr'].map(manufacturers)
```

```python
rating_df.head(3)
```

```python
rating_df = rating_df.join(
    nutrition[['sugar_level', 'protein_level', 'fiber_level']]
  ).copy()
```

```python
rating_df.head()
```

```python
rating_df.groupby('sugar_level')['rating'].agg(
    ['count', 'min', 'mean','max']
).round(2)
```

**Қант мөлшері мен өнім рейтингі арасында қандай байланыс бар?**

Low sugar: қант мөлшері төмен өнімдердің орташа рейтингі ең жоғары (38.84).


High sugar: қант мөлшері жоғары өнімдердің орташа рейтингі ең төмен (18.04).


Moderate sugar: орташа рейтинг 19.82, яғни High санатына жақын.

```python
rating_df.groupby('protein_level')['rating'].agg(
    ['count', 'min', 'mean','max']
).round(2)
```

```python
rating_df.groupby('fiber_level')['rating'].agg(
    ['count', 'min', 'mean','max']).round(2).sort_values('max', ascending=False)
```

**Ақуыз бен тағамдық талшық мөлшері рейтингке қалай байланысты?**

Екі көрсеткіш бойынша да ұқсас тенденция байқалады: ақуыз бен тағамдық талшық мөлшері жоғары өнімдер жоғары рейтинг алуға бейім. Әсіресе тағамдық талшық бойынша айырмашылық айқын көрінеді.

```python
rating_df.query("calories > 50 and fat > 1 and sodium < 100")
```

```python
rating_df.query("calories > 50 and fat > 1 and sodium < 100").sort_values('mfr')
```

**Тағамдық құрамы бойынша ерекшеленетін өнімдердің рейтингі де өзгеше ме?**

Калория, май және натрий мөлшерінің өнім рейтингімен байланысы
calories > 50, fat > 1, sodium < 100 шарттары бойынша іріктелген өнімдердің рейтингтері салыстырмалы түрде төмен болды (33.98–50.83).

Бұл өнімдер жоғары рейтинг алған жармалардан тағамдық құрамы бойынша ерекшеленеді. Атап айтқанда, жоғары рейтингті өнімдердің арасында тағамдық талшық пен ақуыз мөлшері жоғары, қант мөлшері төмен жармалар басым.

# Жалпы қорытынды

Cereals деректер жиынтығына жүргізілген талдау нәтижесінде өндірушілердің өнімдері рейтингі мен тағамдық құрамы бойынша ерекшеленетіні анықталды.

1\. **Өндірушілер рейтингі:** Nabisco өнімдері ең жоғары орташа рейтингке (67.97) ие болды, ал General Mills ең төмен көрсеткіш көрсетті (34.49). Дегенмен, ең жоғары жеке өнім рейтингі Kelloggs өндірушісінің All-Bran with Extra Fiber өніміне тиесілі (93.70).

2\. **Тағамдық құрамы:** өндірушілер арасында қант, ақуыз және тағамдық талшық мөлшері бойынша айырмашылықтар байқалды. Kelloggs өнімдерінің арасында қант мөлшері жоғары жармалар, соның ішінде Smacks кездеседі. Nabisco және Kelloggs өнімдері ақуыз бен тағамдық талшық мөлшері бойынша ерекшеленеді.

3\. **Құрамы және рейтинг:** қант мөлшері төмен өнімдердің орташа рейтингі жоғары (38.84), ал қант мөлшері жоғары өнімдердің рейтингі төмен (18.04) болды. Бұл қант мөлшері мен рейтинг арасында кері байланыстың болуы мүмкін екенін көрсетеді.

Ақуыз және тағамдық талшық: екі көрсеткіш бойынша да мөлшері жоғары өнімдер жоғары рейтингке ие болуға бейім. Әсіресе тағамдық талшық мөлшері бойынша айырмашылық айқын байқалды.

Калория, май және натрий: калориясы 50-ден жоғары, май мөлшері 1 г-нан жоғары және натрий мөлшері 100 мг-нан төмен өнімдердің рейтингтері салыстырмалы түрде төмен болды (33.98–50.83). Бұл өнімдердің тағамдық құрамы жоғары рейтингті жармалардан ерекшеленеді.

**Жалпы тұжырым:** талдау нәтижелері жарма өнімдерінің рейтингі мен тағамдық құрамы арасында белгілі бір байланыс бар екенін көрсетті. Ақуыз бен тағамдық талшық мөлшерінің жоғары болуы және қант мөлшерінің төмен болуы жоғары рейтингпен байланысты болды. Сонымен қатар өндірушілер арасында да өнім сапасы мен тағамдық құрамы бойынша айырмашылықтар байқалды.
